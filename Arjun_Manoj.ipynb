```python
import pandas as pd
```

```python
students=pd.Series([78,85,92,67,74,88,95,61],index=["Anu","Binu","Cathy","Deepa","Ebin","Fathima","Gopal","Hari"]) #1.	Create a series score from the list of scores with student names as the index.
department=pd.Series(["CS","IT","CS","BCA","IT","CS","CS","BCA"])
```

```python
#2.	Create a series dept from a Python dictionary mapping student to department.

nd=dict(zip(students,department))
print(nd)
print(students)
```

```python
#3. 3.	Set the name of scores to "Test Score" and display it.

students.name="Test Score"
print(students)

```

```python
#4.	Display the values, index, data type, and name of scores using the attributes

print(students.values)
print(students.dtype)
print(students.name)
print(students.index)
```

```python
#5.	Display Cathy  score using its label

print(f'Cathy score is {students.loc["Cathy"]}')
```

```python
#6.	Display the first student's score using positional indexing.

print(f'first student score is {students.iloc[0]}')
```

```python
#7.	Display the scores from Binu to Ebin using label slicing, then the first three students using iloc slicing. State which end is excluded in each.

print(f'score from Binu to Ebin is \n {students.loc["Binu":"Ebin"]}')
print(f'scores of first three students is \n {students.iloc[0:3]}')
```

```python
#8.	Display the scores of Anu, Cathy and Ebin using multi-label indexing.

print(f'scores of Anu,Cathy,Ebin is \n {students.iloc[[0,2,4]]}')
```

```python
#9.	Find the highest score, the lowest score, the mean (rounded to 2 decimals) and the number of students.

print(f'highest score is {students.max()}')
print(f'lowest score is {students.min()}')
print(f'mean of scores is {round(students.mean(),2)})')
print(f'number of students is  {students.count()}')
```

```python
#10.	Display the students who scored 80 or above.

print(f'students scored more than 80 is \n {students[students>=80]}')
```

```python
#11.	Display the top 3 scorers in descending order.

s=students.sort_values(ascending=False)
print(f'the top 3 scoresrs is \n {s[:3]}')
```

```python
#12.	Find the name of the student with the highest score
print(f'student with highest score is {students.idxmax()}')
```

```python
#13.	Write a function grade(score) that returns:
#a.	90 and above: "Distinction"
#b.	75 to 89: "First Class"
#c.	60 to 74: "Second Class"
#d.	below 60: "Fail"
#Apply it to scores and store the result in grades.

g=lambda m:"Distinction" if m>90 else "First Class" if (m>=75) & (m<=89) else "Second Class" if(m>=60)& (m<=74) else "Fail"
result=students.apply(g)
print(result)
```

```python
#14.	Display how many students fall in each grade, then the same as percentages.

print(result.value_counts())
print(result.value_counts(normalize=True))
```

```python
#15.	Display the unique grades and the number of unique grades.

print(f'unique grades {result.unique()}')
print(f'number of unique grades is {result.nunique()}')
```

```python
#16.	Create a new Series bonus_scores by adding 5 grace marks to every student, capped at 100. Confirm the original scores is unchanged.

a=lambda d:d+5
bonus_score=students.apply(a)
print(bonus_score)
print(students)
```
