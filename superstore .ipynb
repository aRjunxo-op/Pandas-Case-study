```python

import pandas as pd

df = pd.read_csv('/content/Sample - Superstore.csv', encoding='latin1')
df.head()
```

```python
#What is the size and structure of the dataset

print("Rows and columns:", df.shape)
print("Column names:", df.columns)
df.info()
```

```python
#Missing values and duplicate records

print("Missing values:")
print(df.isnull().sum())

print("Duplicate records:", df.duplicated().sum())
```

```python
#What are the total sales, total profit, and average order-line sales

print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Average Order-Line Sales:", df["Sales"].mean())
```

```python
#Which product categories generate the most sales

category_sales = df.groupby("Category")["Sales"].sum()
category_sales = category_sales.sort_values(ascending=False)
print(category_sales)
```

```python
#Which sub-categories generate the highest and lowest profit

subcategory_profit = df.groupby("Sub-Category")["Profit"].sum()
subcategory_profit = subcategory_profit.sort_values(ascending=False)
print(subcategory_profit)
```

```python
#Which region has the highest sales

region_sales = df.groupby("Region")["Sales"].sum()
region_sales = region_sales.sort_values(ascending=False)
print(region_sales)
```

```python
#Which customer segment contributes the most sales

segment_sales = df.groupby("Segment")["Sales"].sum()
segment_sales = segment_sales.sort_values(ascending=False)
print(segment_sales)
```

```python
#How do discounts relate to profit

discount_profit = df.groupby("Discount")["Profit"].mean()
discount_profit = discount_profit.sort_index()
print(discount_profit)
```

```python
#How do sales change over time

df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
print(monthly_sales)
```
