```python
emp['Age'] = pd.to_numeric(emp['Age'], errors='coerce').astype('Int64')
print(emp['Age'].dtype)
print(emp['Age'].isnull().sum())
```

```python
Q1 = emp['Salary'].quantile(0.25)
Q3 = emp['Salary'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

outliers = emp[(emp['Salary'] < lower_limit) | (emp['Salary'] > upper_limit)]

print(outliers)
```

```python
df_clean = emp[(emp['Salary'] >= lower_limit) & (emp['Salary'] <= upper_limit)]

print("Correlation before removing outlier:", emp['Experience'].corr(emp['Salary']))
print("Correlation after removing outlier:", df_clean['Experience'].corr(df_clean['Salary']))
```

```python
emp['Overall_Score'] = emp[['PerformanceScore', 'Attendance']].mean(axis=1)

emp['Annual_Bonus'] = emp['Salary'] * 0.10

print(emp[['Overall_Score', 'Annual_Bonus']])
```

```python
emp['Grade'] = emp['PerformanceScore'].apply(
    lambda x: 'Excellent' if x >= 85 else 'Good' if x >= 70 else 'Needs Improvement'
)

print(emp['Grade'].value_counts())
print(emp.groupby('Grade')['Salary'].mean())
```
