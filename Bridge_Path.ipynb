```python
import pandas as pd
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

revenue_values = [42, 45, 50, 48, 55, 60, 58, 62, 65, 70, 75, 85]

expense_values = [38, 40, 42, 47, 50, 52, 55, 60, 61, 64, 66, 72]
```

```python
#Create Revenue and Expenses Series

revenue = pd.Series(revenue_values, index=months, name="Revenue")
expenses = pd.Series(expense_values, index=months, name="Expenses")
print(revenue)
print(expenses)
```

```python
#Python Lists Addition
a = [42, 45, 50]
b = [38, 40, 42]
print(a + b)
```

```python
#Calculate Profit Using a For Loop

profit_list = []
for i in range(len(revenue_values)):
    p = revenue_values[i] - expense_values[i]
    profit_list.append(p)
print(profit_list)
```

```python
#Calculate Profit Using a Pandas Series

profit_series = revenue - expenses
print(profit_series)
print("Matches A3:", profit_series.tolist() == profit_list)
```

```python
#Create the Profit Series

profit = revenue - expenses
profit.name = "Profit"
print(profit)
```

```python
#Calculate Profit Margin

margin = (profit / revenue * 100).round(1)
print(margin)
```

```python
#Calculate Yearly Financial Totals

total_revenue = revenue.sum()
total_expenses = expenses.sum()
total_profit = profit.sum()
overall_margin = (total_profit / total_revenue * 100).round(1)
print("Total Revenue:", total_revenue, "lakh")
print("Total Expenses:", total_expenses, "lakh")
print("Total Profit:", total_profit, "lakh")
print("Overall Profit Margin:", overall_margin, "%")
```

```python
#Quarterly Profit Analysis Using loc

Q1 = profit.loc["Jan":"Mar"].sum()
Q2 = profit.loc["Apr":"Jun"].sum()
Q3 = profit.loc["Jul":"Sep"].sum()
Q4 = profit.loc["Oct":"Dec"].sum()

print("Q1 Profit:", Q1, "lakh")
print("Q2 Profit:", Q2, "lakh")
print("Q3 Profit:", Q3, "lakh")
print("Q4 Profit:", Q4, "lakh")

quarter_profits = pd.Series({
    "Q1": Q1,
    "Q2": Q2,
    "Q3": Q3,
    "Q4": Q4
})

print("Strongest Quarter:", quarter_profits.idxmax())
```

```python
#First-Half and Second-Half Profit Using iloc

H1 = profit.iloc[0:6].sum()
H2 = profit.iloc[6:12].sum()
print("First Half (H1) Profit:", H1, "lakh")
print("Second Half (H2) Profit:", H2, "lakh")
```

```python
#Display Festive Months Profit

festive_profit = profit.loc[["Oct", "Nov", "Dec"]]
print(festive_profit)
```

```python
print(profit.iloc[3:6])
```

```python
#Display Months Where Profit Margin Exceeds 10%

print(margin[margin > 10])
```

```python
#Display Months Where Profit Is Below ₹4 Lakh

print(profit[profit < 4])
```

```python
#Find the Best and Worst Months

best_profit = profit.max()
worst_profit = profit.min()
print("Best month:")
print(profit[profit == best_profit])
print("Worst month:")
print(profit[profit == worst_profit])
```

```python
#Display the Top 3 Months by Profit

top_three = profit.sort_values(ascending=False).head(3)
print(top_three)
```

```python
#Average Monthly Profit

average_profit = profit.mean()
print("Average Monthly Profit:", round(average_profit, 2), "lakh")
print("Months exceeding average profit:")
print(profit[profit > average_profit])
```

```python
#Create the profit_status() Function

def profit_status(p):
    if p >= 8:
        return "Strong"
    elif p >= 4:
        return "Stable"
    else:
        return "Weak"

status = profit.apply(profit_status)
print(status)
```

```python
#Number and Percentage of Months in Each Status

status_counts = status.value_counts()
status_percentages = (status.value_counts(normalize=True) * 100).round(1)
print("Number of months in each status:")
print(status_counts)
print("\nPercentage of months in each status:")
print(status_percentages)
```

```python
#Calculate Tax Using tax_payable()

def tax_payable(p):
    if p > 0:
        return p * 0.25
    else:
        return 0
tax = profit.apply(tax_payable)
print(tax)
```

```python
#Calculate Total Tax and Post-Tax Profit

total_tax = tax.sum()
post_tax_profit = profit.sum() - total_tax
print("Total Tax:", round(total_tax, 2), "lakh")
print("Post-Tax Profit:", round(post_tax_profit, 2), "lakh")
```
