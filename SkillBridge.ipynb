```python
#Create a Series for study modes

study_modes = pd.Series(mode_list, name="Study Mode")
print(study_modes)
```

```python
#Display the number of learners and percentage for each study mode

mode_counts = study_modes.value_counts()
mode_percentages = (study_modes.value_counts(normalize=True) * 100).round(1)
print("Number of learners:")
print(mode_counts)
print("\nPercentage share:")
print(mode_percentages)
```

```python
#Find the most preferred study mode and the combined percentage of Online and Hybrid.

most_preferred_mode = mode_counts.idxmax()
online_hybrid_percentage = (mode_percentages.loc[["Online", "Hybrid"]].sum()).round(1)
print("Most preferred study mode:", most_preferred_mode)
print("Combined Online and Hybrid percentage:", online_hybrid_percentage, "%")
```

```python
#Create a data structure named ratings from the rating list. Display the number of ratings, the average rating (rounded to 2 decimal places), the highest rating, and the lowest rating.

ratings = pd.Series(rating_list)

print("Total ratings:", len(ratings))
print("Average rating:", round(ratings.mean(), 2))
print("Highest rating:", ratings.max())
print("Lowest rating:", ratings.min())
```

```python
#Count each rating score

print(ratings.value_counts().sort_index())
```

```python
def sentiment(r):
    if r == 4 or r == 5:
        return "Positive"
    elif r == 3:
        return "Neutral"
    else:
        return "Negative"
sentiments = ratings.apply(sentiment)
print(sentiments)
```

```python
#Display the percentage of learners in each sentiment category, rounded to 1 decimal place.

sentiment_percentages = (sentiments.value_counts(normalize=True) * 100).round(1)
print(sentiment_percentages)
```
