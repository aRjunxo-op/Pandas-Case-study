```python
missing_report = pd.DataFrame({
    'Missing Count': hs.isnull().sum(),
    'Missing Percentage': (hs.isnull().sum() / len(hs) * 100).round(2)
})

print(missing_report)

hs['BillAmount'] = hs['BillAmount'].fillna(hs['BillAmount'].median())
hs['SatisfactionScore'] = hs['SatisfactionScore'].fillna(hs['SatisfactionScore'].median())
```

```python
print(hs[hs.duplicated(keep=False)])

print(hs.duplicated().sum())

hs = hs.drop_duplicates()

print(hs['PatientID'].is_unique)
```

```python
hs['Age'] = pd.to_numeric(hs['Age'].replace('unknown', np.nan), errors='coerce')

print(hs.isnull().sum())

median_age = hs['Age'].median()
hs['Age'] = hs['Age'].fillna(median_age).astype(int)
```

```python
hs['AdmitDate'] = pd.to_datetime(hs['AdmitDate'], errors='coerce')

print("Invalid dates:", hs['AdmitDate'].isna().sum())

for col in ['Department', 'Gender', 'Insurance']:
    hs[col] = hs[col].astype('string')

print(hs.dtypes)
```

```python
print("StayDays vs BillAmount:", hs['StayDays'].corr(hs['BillAmount']))

print("StayDays vs SatisfactionScore:", hs['StayDays'].corr(hs['SatisfactionScore']))

print("Age vs StayDays:", hs['Age'].corr(hs['StayDays']))
```

```python
q1 = hs['BillAmount'].quantile(0.25)
q3 = hs['BillAmount'].quantile(0.75)

iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

outliers = hs[(hs['BillAmount'] < lower_limit) | (hs['BillAmount'] > upper_limit)]

print(outliers)
```

```python
print(hs['StayDays'].corr(hs['BillAmount']))
```

```python
hs_clean = hs[(hs['BillAmount'] >= lower_limit) & (hs['BillAmount'] <= upper_limit)].copy()

print(hs_clean['StayDays'].corr(hs_clean['BillAmount']))
```

```python
hs_clean['Cost_Per_Day'] = hs_clean['BillAmount'] / hs_clean['StayDays']

hs_clean.groupby('Department')['Cost_Per_Day'].mean().sort_values(ascending=False)
```

```python
hs_clean['Stay_Type'] = hs_clean['StayDays'].apply(
    lambda x: 'Short' if x <= 3 else 'Medium' if x <= 7 else 'Long'
)

hs_clean['Age_Group'] = hs_clean['Age'].apply(
    lambda x: 'Child' if x < 18 else 'Adult' if x < 60 else 'Senior'
)

print(hs_clean['Stay_Type'].value_counts())
print(hs_clean['Age_Group'].value_counts())
```

```python
hs_clean.groupby('Stay_Type')['SatisfactionScore'].mean()
```

```python
hs_clean.groupby('Age_Group')['SatisfactionScore'].mean()
```

```python
hs_clean.to_csv('cleaned_hospital.csv', index=False)
```
